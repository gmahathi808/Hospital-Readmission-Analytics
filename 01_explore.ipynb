```python
import pandas as pd
pd.set_option("display.max_columns", None)

df = pd.read_csv("../data/raw/diabetic_data.csv", na_values="?", low_memory=False)
df.shape
```

```python
df.isna().sum()[lambda s: s > 0]
```

```python
df["encounter_id"].duplicated().sum()
```

```python
dead_or_hospice = [11, 13, 14, 19, 20, 21]
print("Rows to remove:", df["discharge_disposition_id"].isin(dead_or_hospice).sum())

df = df[~df["discharge_disposition_id"].isin(dead_or_hospice)]
df.shape
```

```python
print(df["gender"].value_counts())
df = df[df["gender"] != "Unknown/Invalid"]
df.shape
```

```python
print("Columns with only one value:", [c for c in df.columns if df[c].nunique() == 1])

df = df.drop(columns=["weight", "examide", "citoglipton"])
df.shape
```

```python
for col in ["race", "medical_specialty", "payer_code"]:
    df[col] = df[col].fillna("Unknown")

for col in ["A1Cresult", "max_glu_serum"]:
    df[col] = df[col].fillna("Not tested")

df["A1Cresult"].value_counts()
```

```python
df["readmitted_30d"] = (df["readmitted"] == "<30").astype(int)

print(df["readmitted_30d"].value_counts())
print("30-day readmission rate:", round(df["readmitted_30d"].mean() * 100, 2), "%")
```

```python
df["age_years"] = df["age"].str.extract(r"\[(\d+)-")[0].astype(int) + 5
df[["age", "age_years"]].drop_duplicates().sort_values("age_years")
```

```python
import os
os.makedirs("../data/processed", exist_ok=True)      # create the folder if it isn't there
df.to_csv("../data/processed/encounters_clean_step2.csv", index=False)
```
